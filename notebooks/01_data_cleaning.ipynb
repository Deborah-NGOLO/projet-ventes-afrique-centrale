| cout_fcfa | Coût total associé à la vente |

| methode_paiement | Moyen de paiement utilisé |
| campagne_marketing | Campagne marketing associée |
| satisfaction_client | Niveau de satisfaction du client |

8. Suppression de la variable order_id

```python
df=df.drop('order_id',axis=1)
```

**9. Renommage de colonnes**

```python
df.columns = [
    "date_commande",
    "pays",
    "ville",
    "canal_vente",
    "segment_client",
    "categorie_produit",
    "produit",
    "quantite",
    "prix_unitaire_fcfa",
    "remise_pourcentage",
    "chiffre_affaires_fcfa",
    "cout_total_fcfa",
    "profit_net_fcfa",
    "methode_paiement",
    "campagne_marketing",
    "satisfaction_client"
]
```

**10. Vérification des noms de colonnes**

```python
df.head()
```

**11. Conversion de la date**

```python
df["date_commande"] = pd.to_datetime(df["date_commande"])
```

12.Création des variables temporelles

```python
df["annee"] = df["date_commande"].dt.year
df["mois"] = df["date_commande"].dt.month
```
